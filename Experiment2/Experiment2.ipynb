```python
# Import
import numpy as np
import pandas as pd
```

```python
# Load the dataset

cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']
df = pd.read_csv("iris.csv", header=None, names=cols)

print(df)

print(df.shape)
```

```python
# Calculate Central tendencies
mean_val=df['sepal_length'].mean()
median_val=df['sepal_length'].median()
mode_val=df['sepal_length'].mode()

print("Mean value :",mean_val)
print("Median value :",median_val)
print("Mode value :",mode_val)
```

```python
# Calculate Dispersion Measures

## Calculate Variance
variance_val = df['sepal_length'].var()

## Calculate Standard Deviation
std_dev_val = df['sepal_length'].std()

print("Variance value :",variance_val)
print("standard deviation :",std_dev_val)

# Calculate quartiles
quartiles = df['sepal_length'].quantile([0.25, 0.50, 0.75])

q1 = quartiles[0.25]
q2 = quartiles[0.50]  
q3 = quartiles[0.75]
print(q1,q2,q3)

# Calculate IQR
IQR = q3 - q1

print(IQR)
```
